## Caesar Cipher

<div class="row" style="">
    <div class="column" style="text-align:center">
        <img src="./img/history/caesar-cipher-wheel-.webp" style="vertical-align:middle;width:30%" >
</div>

* In the Caesar cipher every letter is shifted in the alphabet by $k$
* Named after *Julius Caesar*, who is said to have shifted by $3$
  + `ROT13` is Caesar cipher with $k = 13$
* **Substitution** cipher: Every letter is replaced

### Caesar Cipher

```python
def caesar(plain, k):
  ROT = ( lambda c: chr((ord(c) - 65 + k) % 26 + 65)  
                    if c.isalpha() else c )
  return "".join(ROT(c) for c in plain.upper())
    
c = caesar("Attack at dawn", 3); 
print(f"ciphertext = {c}"); 
print(f"plaintext  = {caesar(c, -3)}")
```

### Breaking the Ceasar Cipher

* Key space: 25 keys
    + Brute force by hand, over coffee

**Lession:**
<div style="background-color:#ffcccc; color:darkred; padding:10px 14px; border-left:5px solid red; border-radius:4px">
A large key space is required for any secure cipher. 
</div>

#### Breaking the Ceasar Cipher 

```python
def caesar_break(cipher):
    for k in range(0,27):
        print(caesar(cipher,-k))
caesar_break(c)
```

#### Breaking the Ceasar Cipher 

Find the (most likely) correct english plaintext:

```python
EN = "ETAOINSHRDLCUMWFGYPBVKJXQZ" # letters sorted by English frequency
# score by sum of indices (raning from 0-26)
score = lambda s: sum(EN.index(ch) for ch in s if ch.isalpha())
# smallest sum of indices wins => most likely english text
best_key = min(range(26), key=lambda k: score(caesar(c, -k)))
print("key:", best_key, "->", caesar(c, -best_key))
```

### Breaking the Ceasar Cipher (without brute force)

* Letter frequencies survive substitution: `E` stays the most frequent in english

* Allow *any* letter permutation as key: $26! \approx 2^{88}$ keys, but letter frequencies remain 

```python
math.log2(math.factorial(26))
```

**Lesson:** 
<div style="background-color:#ffcccc; color:darkred; padding:10px 14px; border-left:5px solid red; border-radius:4px">
A large key space is necessary, but does <span style="font-style:italic;">not</span> automatically make a cipher strong.
</div>

#### Breaking the Ceasar Cipher (without brute force)

```python
c = caesar("the enemy knows the system", 7)
ctr = Counter(c for c in ct if c.isalpha()).most_common(3)
best_chr = min([ch for ch, _ in ctr], key=lambda k: score(caesar(c, -abs(ord("E") - ord(k)))))
best_key = abs( ord("E") - ord(best_chr) )
print(ctr); print("key:", best_key, "->", caesar(c, -best_key))
```

## One-Time-Pad (OTP)

* Vernam 1917; **Shannon 1949**: proven unbreakable
* $c = m \oplus k$ &nbsp;and&nbsp; $m = c \oplus k$
* Three rules: the key is **truly random**, **as long as** the message, used **exactly once**
* **Perfect secrecy**: the ciphertext tells Eve *nothing*, every plaintext of that length stays equally likely, forever

```python
xor = lambda a, b: bytes(x ^ y for x, y in zip(a, b)) # char wise xor

m, k = b"ATTACK AT DAWN", os.urandom(14)   # key: random, |k| = |m|, used ONCE
print(xor(m, k).hex())                     # c = m XOR k
print(xor(xor(m, k), k))                   # m = c XOR k
```

## Breaking OTP

* You must ship **as much key material as you will ever send** &mdash; securely, in advance
* **No integrity**: flip a bit in $c$ and the same bit flips in $m$, undetected
* Use the key twice and it collapses: $c_1 \oplus c_2 = m_1 \oplus m_2$ &mdash; the key is gone
    + **VENONA**: reused Soviet pads let the US read traffic for decades

```python
m1, m2 = b"KNOWN MESSAGE1", b"ATTACK AT DAWN"
c1, c2 = xor(m1, k), xor(m2, k)                        # same key twice -- don't
print(xor(c1, c2) == xor(m1, m2))                      # the key cancels out
print(bytes(x ^ y ^ z for x, y, z in zip(c1, c2, m1)))  # guess m1 -> read m2
```

## Principles of modern cryptography

* <span style="color:blue; font-weight:bold;">**Precise Assumptions**</span>: Assumptions are statements that are not proven but are instead
conjectured to be true
  + assumption about the hardness of a mathematical problem e.g., factoring, discreate logarithm, AES $\approx$ random permutation, ...
  + existance of one way functions, requires the assumption $\mathtt{P} \neq \mathtt{NP}$

* <span style="color:blueviolet; font-weight:bold;">**Formal Definitions**</span>: The **model** defines what threats are in scope and what security guarantees are desired
    + **cryptographic goals** (what should be achieved)
        + e.g., in context of encryption: *a ciphertext should leak no additional information about the underlying plaintext*
    + **attacker model** (power of the attacker)
      + e.g., in context of encryption: ciphertext-onlyc (COA), known-plaintext (KPA), ...

* <span style="color:darkgreen; font-weight:bold;">**Proofs of Security**</span>:  Proof that a construction satisfies a given definition under a certain attacker model and assumptions
    + by reduction *"if an attacker can break my scheme, he would also be able to break the assumption"*

## Cryptographic Goals 

* **Confidentiality**, nobody else reads it &rarr; encryption
* **Integrity**, nobody changed it &rarr; hash, MAC, signature
* **Authenticity**, it really came from Bob &rarr; MAC, signature
* **Non-repudiation**, Bob cannot deny it later &rarr; signature only
* **Repudiation**, Bob can deny it later, &rarr; designated verifyer 
* **Freshness**, no replay &rarr; incl. challenge
* **Forward secrecy**, yesterday's traffic stays safe &rarr; DH
* ...
  
*Availability is a security goal, but not a cryptographic one.*


More examples:
* **Post compromise security** eventually future conversations will be secure again
* **Zero knowlege**, verifyer only learns that the statement is true
* **Designated verifyer**, Bob (verifyer) is not able to convince a third party

## Attacker Models

Most important ones in for encryption, from low- to high attacker capabilities:

|         |                 |           |
|---------|-----------------|-----------|
| **COA** | ciphertext only | Attacker has only access to ciphertexts |
| **KPA** | known plaintext attack | Attacker knows plaintext &lrarr; ciphertext paris |
| **CPA** | chosen plaintext attack | Attacker can choose plaintexts to get encrypted as well|
| **CCA** | chosen ciphertext attack | Attacker can choose ciphertexts to be decrypted, before he gets to know the ciphertext he is interested in (non-adaptive aka. *lunchtime attack*) |
| **CCA2** | adaptive chosen ciphertext attack | Attacker can choose some ciphertexts to be decrypted, before and after he gets to know the ciphertext he is interested in (adaptive). The attacker cannot use the oracle to decrypt the ciphertext he is interested in|

* Types of **break**: key recovery &sub; plaintext recovery &sub; **distinguish** encrypted plaintexts 
* Important bar: **IND-CPA** *indistinguishability* under a chosen plaintext attacker
  + Encryption must be *randomized* &rArr; IV / nonce
* Outside the model: *side channels* (timing, power, cache, faults etc.), *implementaiton* issues, *usage/user*, ...

### From Information Theoretical Security to Computational Security

* The one-time pad provides perfect secrecy but **requires** $|k| \ge |m|$ (Shannon)

* So we relax the *attacker*, not the goal:
    + he runs in **reasonable time** (polynomial), not forever
    + he may win with **negligible** probability

* Assume **one-way functions** exist ($\text{P} \neq \text{NP}$) and the whole *symmetric* world follows:
    + **OWF** &rarr; **PRG** &rarr; stream cipher &nbsp;&middot;&nbsp; **PRF / PRP** &rarr; block cipher, MAC

* Assume certain math problems are hard, **trapdoor functions** exist and the whole *asymmetric* world follows:
    + **factoring** &rarr; RSA problem &rarr; RSA &nbsp;&middot;&nbsp; **Dlog** &rarr; DDH/CDH &rarr; DH &nbsp;&middot;&nbsp; **LWE** &rarr; Lattice based

### Measure Security

* "*128-bit security*" = the best *known* attack costs about $2^{128}$ operations
    + A super computer that can try $2^{90}$ keys per day would need millions of years to guess a 128-bit key
    + no budget, no hardware and no patience gets you there
