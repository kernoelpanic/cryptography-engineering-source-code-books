## The Public-Key Revolution (1976)
<div style="display:flex; align-items:flex-start; gap:1em">

<div style="flex:6">
    
* **DH** 1976 (*Whitfield Diffie, Martin Hellman*): ["New Directions in Cryptography"](https://www.cs.jhu.edu/~rubin/courses/sp03/papers/diffie.hellman.pdf): allowed to compute a shared secret over public channel
    + **public key** $pk$ &nbsp; hand it to everyone
    + **private key** $sk$ &nbsp; keep secret
* **RSA** 1977 (Rivest, Shamir, Adleman): first practical public-key encryption *and* signature scheme
* Declassified later: Ellis, Cocks and Williamson had the idea at GCHQ in 1969&ndash;73
* All of it rests on **trapdoor** problems: easy one way, hard back
    + factoring, discrete log (mod $p$ and on elliptic curves), lattices, ...

### Diffie&ndash;Hellman Key Exchange 

* **Public parameters** (PP): a large prime $p$ and a generator $g$ (of a cyclic sub-group!)

* Alice picks a secret $a$ and sends $A = g^a \bmod p$
* Bob picks a secret $b$ and sends $B = g^b \bmod p$
* Both compute the shared secret $B^a = A^b = g^{ab} \bmod p$ 

```python
def DH(g,x,p): 
    return pow(g,x,p) # g^x mod p
```

* Eve sees $g^a$ and $g^b$ and is stuck on the **discrete log problem** (Dlog)
* Mallory could launch *Mashine-in-the-middle* (MITM) attack

* [Toy example with clocks](./dh_clock.html)

### Diffie&ndash;Hellman Key Exchange

```python
p, g = 17, 3       # toy parameters! real: >= 3072-bit prime or EC
a, b = 5, 9        # Alice's secret, Bob's secret
A, B = DH(g, a, p), DH(g, b, p)  # what goes over the wire
print(DH(B, a, p) == DH(A, b, p), DH(B, a, p))
```

### Caution &#9888; 
Just a toy example, to illustrate the concept, not for production use! 
+ Non-secure public parameters can be used (e.g., to small)
+ Does not validate inputs (operate with secret key on invalid parameters)

```python
invalid_pub_key = DH(3, a, 2**1024) # Bob forces invalid PP (large) p 
print(f"invalid pub. key A = {invalid_pub_key}") 
print(f"secret of Alice  a = {math.ceil(math.log(invalid_pub_key,3))}")
```

## RSA

* **Public key**: $(n, e)$
    + $n = p \cdot q$ where $p, q$ are large primes
* **Private key**: $d,p,q$
    + where $m^{ed} \equiv m \pmod n$
* Security rests on factoring $n$ &rarr; 2048-bit minimum, 3072+ for anything new

```python
def rsa_enc(n,e,m): 
	return pow(m,e,n) # m^e mod n 
def rsa_dec(n,d,c):
	return pow(c,d,n) # c^d mod n
```

### Caution &#9888; 
* **Textbook RSA is insecure**: deterministic, malleable, small-message attacks
    + never encrypt your payload with it directly
    + encryption &rarr; **RSA-OAEP** &nbsp;&middot;&nbsp; signatures &rarr; **RSA-PSS**
      (no PKCS#1 v1.5 in new code)

## Hybrid Encryption / KEM & DEM

* Asymmetric crypto to negotiate **key**, symmetric crypto for the **data** (faster)
    + **KEM** (key encapsulation mechanism) **DEM** (data encapsulation mechanism)
    + **KEM**: encapsulate / agree on a fresh session key &rarr; **DEM**: AEAD over the payload
* That is TLS, SSH, `gpg`, Signal, `age`, disk and backup encryption, ...
* Which leaves the hard part: *whose public key is this?*
    + **PKI** and X.509 certificates, Certificate Transparency, web of trust, TOFU (SSH)
    + An unauthenticated public key could be just some attacker's key
* Use **X25519** (hard to misuse) as KEM

## Digital Signatures

* Sign with $sk$, everyone verifies with $vk$ (verification key)
  + in this context the public key is called *verification key* 
  + integrity + authenticity + **non-repudiation**
* Always *hash-then-sign*: sign $H(m)$, not $m$
* Signatures are what certificates, software updates, eIDs and blockchains are built on
* Use **Ed25519** (hard to misuse), ECDSA (P-256), or RSA-PSS
    + ECDSA footgun: a reused or biased nonce **leaks the private key**
